```python
# -*- coding: utf-8 -*-
"""
Este notebook demonstra a conversão de voz em texto usando Python e como usar o texto resultante
para interagir com diferentes sites, como Wikipedia.
"""
# =================================================================================================
# 1. Configuração e Importação de Bibliotecas
# =================================================================================================

# Para conversão de voz para texto, usaremos a biblioteca SpeechRecognition.
# Para lidar com o formato de áudio (se necessário), usaremos pydub.
# Para pesquisar na Wikipedia, usaremos a biblioteca 'wikipedia'.
%pip install SpeechRecognition pydub wikipedia requests

import speech_recognition as sr
import os
import pydub
import wikipedia
import requests
from bs4 import BeautifulSoup
import warnings

# Ignorar avisos da Wikipedia que podem aparecer
warnings.filterwarnings("ignore", category=UserWarning)

# Instalar dependências necessárias para o PyAudio no ambiente do Colab
%apt-get install -y portaudio19-dev
%pip install pyaudio

# =================================================================================================
# 2. Captura de Áudio e Conversão para Texto
# =================================================================================================

# Nota: A gravação direta pelo microfone em um ambiente de notebook pode ser complexa.
# A melhor abordagem para este notebook é usar um arquivo de áudio pré-existente
# ou simular a gravação. Abaixo, temos um exemplo de como gravar um arquivo de áudio,
# mas para que o notebook seja completamente executável, vamos assumir que o arquivo de áudio já existe
# ou criaremos um de exemplo.

def record_audio(filename="audio_input.wav"):
    """
    Grava áudio do microfone e salva em um arquivo WAV.
    (Esta função pode não funcionar em todos os ambientes de notebook)
    """
    r = sr.Recognizer()
    mic = sr.Microphone()
    print("Por favor, fale algo...")
    try:
        with mic as source:
            r.adjust_for_ambient_noise(source)
            audio = r.listen(source)
        with open(filename, "wb") as f:
            f.write(audio.get_wav_data())
        print(f"Áudio gravado e salvo em {filename}")
        return filename
    except Exception as e:
        print(f"Não foi possível gravar o áudio: {e}")
        return None

def convert_audio_to_text(audio_file):
    """
    Converte um arquivo de áudio em texto usando a API de reconhecimento de fala do Google.
    """
    if not os.path.exists(audio_file):
        print(f"Erro: O arquivo de áudio '{audio_file}' não foi encontrado.")
        return None

    r = sr.Recognizer()
    try:
        with sr.AudioFile(audio_file) as source:
            audio_data = r.record(source)
            text = r.recognize_google(audio_data, language='pt-BR')
            return text

    except sr.UnknownValueError:
        print("A API de reconhecimento de fala do Google não conseguiu entender o áudio.")
        return None
    except sr.RequestError as e:
        print(f"Erro na requisição da API de reconhecimento de fala do Google; {e}")
        return None

# --- Execução da Conversão ---
# Substitua 'audio_exemplo.wav' pelo caminho do seu arquivo de áudio.
audio_file_path = "audio_input.wav"

# O código agora irá gravar do seu microfone. Certifique-se de permitir o acesso.
audio_file = record_audio(audio_file_path)
if audio_file:
    recognized_text = convert_audio_to_text(audio_file)
    print(f"Texto reconhecido: '{recognized_text}'")
else:
    recognized_text = None

if recognized_text:
    # =================================================================================================
    # 3. Acessar Sites com o Texto Resultante
    # =================================================================================================

    # --- Acessando a Wikipedia ---
    print("\n--- Buscando na Wikipedia ---")
    try:
        # Definir o idioma da pesquisa
        wikipedia.set_lang("pt")

        # Obter o resultado da busca na Wikipedia
        search_query = recognized_text.replace("O que é ", "").strip()
        print(f"Termo de busca na Wikipedia: '{search_query}'")

        # A biblioteca 'wikipedia' é uma forma simples de pesquisar
        page_summary = wikipedia.summary(search_query, sentences=3)
        print("Resumo da Wikipedia:")
        print(page_summary)

    except wikipedia.exceptions.PageError:
        print(f"Não foi possível encontrar uma página da Wikipedia para '{search_query}'.")
    except wikipedia.exceptions.DisambiguationError as e:
        print(f"O termo '{search_query}' é ambíguo. Tente ser mais específico. Opções: {e.options}")

    # --- Acessando Facebook e Portas Abertas ---
    print("\n--- Acessando Facebook e Portas Abertas ---")
    print("Sites como Facebook e Portas Abertas requerem autenticação e têm políticas de uso que impedem o acesso direto e programático.")
    print("O web scraping simples não é viável para essas plataformas. A melhor abordagem é usar a API oficial, se disponível, ou pesquisar por informações públicas através de um motor de busca.")

    # Exemplo de como pesquisar no Google por "Portas Abertas" usando o texto reconhecido
    search_query_facebook = f"site:facebook.com {recognized_text}"
    search_query_portas_abertas = f"site:portasabertas.org.br {recognized_text}"

    print(f"\nExemplo de pesquisa no Google para o Facebook: {search_query_facebook}")
    print(f"Exemplo de pesquisa no Google para Portas Abertas: {search_query_portas_abertas}")

    # Para fazer uma pesquisa real em um buscador, seria necessário usar uma API de busca (como a do Google),
    # que não está disponível neste ambiente. O código acima serve apenas como demonstração conceitual.

else:
    print("\nNão foi possível obter texto para realizar a busca.")
```
