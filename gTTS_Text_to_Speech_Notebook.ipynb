# 🎤 Conversão de Texto em Voz com gTTS

Este notebook converte texto em voz usando a biblioteca **gTTS**.

```
!pip install gTTS
```

```
from gtts import gTTS
from IPython.display import Audio, display

texto = "Olá! Este é um teste de conversão de texto em voz usando gTTS."
tts = gTTS(texto, lang="pt", tld="com.br")
tts.save("voz.mp3")

# Reproduzir áudio no notebook
display(Audio("voz.mp3", autoplay=True))
```

## 🔹 Suporte a Textos Longos

```
def text_to_speech_long(texto, filename="voz_longa.mp3"):
    from gtts import gTTS
    
    # divide em blocos menores
    partes = [texto[i:i+200] for i in range(0, len(texto), 200)]
    tts = gTTS(" ".join(partes), lang="pt", tld="com.br")
    tts.save(filename)
    return filename

# Exemplo
texto_grande = "Este é um exemplo de conversão de um texto longo em áudio com gTTS. " * 10
arquivo = text_to_speech_long(texto_grande)
display(Audio(arquivo, autoplay=True))
```
